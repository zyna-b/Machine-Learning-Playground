```python
import pandas as pd
import seaborn as sns
from sklearn.datasets import make_classification
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
from sklearn.ensemble import RandomForestClassifier
```

```python
X, y = make_classification(n_samples=10, n_classes=2, n_features=2, n_informative=2, n_redundant=0, random_state=42)
```

# Using Decision Trees

```python
clf = DecisionTreeClassifier()
clf.fit(X, y)
```

```python
plot_tree(clf)
```

```python
clf.feature_importances_
```

# Using Random Forest

```python
rfc = RandomForestClassifier(n_estimators=2)
rfc.fit(X, y)
```

```python
rfc.feature_importances_
```

It shows internally our Random forest classifier used 2 Decision trees.

```python
rfc.estimators_
```

Internally Random forest uses seperate decision trees to calculate the feature importance. At the end, for the final result, it just averages out the imporatance of all the decision trees.

```python
print(rfc.estimators_[0].feature_importances_) # Importance of features according to the 1st Decision Tree
print(rfc.estimators_[1].feature_importances_) # Importance of features according to the 2nd Decision Tree

print((1 + 0.555)/2) # Average importance of 1st feature in our data.
```
